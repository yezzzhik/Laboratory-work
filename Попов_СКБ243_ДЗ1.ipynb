```python
import sympy as sp


I1, I2, I3, I4, I5, I6, I7, I8, I9 = sp.symbols('I1 I2 I3 I4 I5 I6 I7 I8 I9')
R1, R2, R3, R4, R5, R6, R7 = 200, 400, 250, 500, 300, 600, 250
E1, E2, E3 = 30, 60, 10

# Контуры
eq1 = I3*R3 - I4*R4
eq2 = I1*R1 + I3*R3 + I5*R5 + I6*R6 - (E1 + E3)
eq3 = I1*R1 + I4*R4 + I5*R5 + I7*R7 - (E1 + E2 + E3)
eq4 = -I1*R1 + I2*R2 + E1

# Узлы
eq5 = -I3 + I6 + I8 - I9
eq6 = I5 - I6 - I7
eq7 = -I4 + I7 - I8
eq8 = I1 + I2 - I6 - I7 + I9
# Источник тока
eq9 = I9 - 1

solution = sp.solve(
    [eq1, eq2, eq3, eq4, eq5, eq6, eq7, eq8, eq9],
    [I1, I2, I3, I4, I5, I6, I7, I8, I9],
    dict=True
)
print("Решение системы:\n")

if solution:
    sol = solution[0]
    for var in [I1, I2, I3, I4, I5, I6, I7, I8, I9]:
        print(f"{var} = {sol[var].evalf():.6f} A")
else:
    print("Система не имеет решений.")
```
