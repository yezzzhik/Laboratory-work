```python
import numpy as np

f = 1591549
omega = 2*np.pi*f

R1 = 100; R2 = 400; R3 = 40; R4 = 800; R5 = 300
C1 = 5e-6; C2 = 30e-6; L1 = 25e-6; L2 = 10e-6

e = 20 * np.exp(1j * np.radians(60))
J = -1 + 0j   

YC1 = 1j*omega*C1
YC2 = 1j*omega*C2
YL1 = 1/(1j*omega*L1)
YL2 = 1/(1j*omega*L2)

Y1 = 1/R1; Y2 = 1/R2; Y3 = 1/R3; Y4 = 1/R4; Y5 = 1/R5

phi3 = e
phi4 = 0+0j

A = np.array([
    [YC2+YL2+Y2,   -YC2,         0,          -Y2],
    [-YC2,         YC2+Y5+Y4,    0,           0 ],
    [0,            0,            YL1+Y3+Y1,  -Y1],
    [-Y2,          0,           -Y1,         Y2+YC1+Y1],
], dtype=complex)

b = np.array([
    -J + YL2*phi3,
    Y4*phi3,
    Y3*phi3,
    YC1*phi3,
], dtype=complex)

phi1, phi2, phi5, phi6 = np.linalg.solve(A, b)

V_n001 = phi1
V_n002 = phi6 
V_n003 = phi3
V_n004 = phi4
V_n005 = phi5


Iv = -(YL2*(phi3 - phi1) +
       Y4*(phi3 - phi2) +
       Y3*(phi3 - phi5) +
       YC1*(phi3 - phi6))


def fmt(z):
    mag = np.abs(z)
    ph = np.degrees(np.angle(z))
    if ph > 180: ph -= 360
    if ph <= -180: ph += 360
    return mag, ph

def print_ltp(name, z, unit=""):
    mag, ph = fmt(z)
    print(f"{name}:\tmag: {mag:10.6g} phase: {ph:10.4f}°\t{unit}")


print_ltp("V(n001)", V_n001, "V")
print_ltp("V(n002)", V_n002, "V")
print_ltp("V(n003)", V_n003, "V")
print_ltp("V(n004)", V_n004, "V")
print_ltp("V(n005)", V_n005, "V")

print_ltp("I(V1)", Iv, "A")

currents = {
    "I(R1)": (phi6-phi5)/R1,
    "I(R2)": (phi1-phi6)/R2,
    "I(R3)": (phi3-phi5)/R3,
    "I(R4)": (phi2-phi3)/R4,
    "I(R5)": (phi2-phi4)/R5,

    "I(C1)": (phi3-phi6)*YC1,
    "I(C2)": (phi1-phi2)*YC2,

    "I(L1)": (phi5-phi4)*YL1,
    "I(L2)": (phi1-phi3)*YL2,
}

for k, v in currents.items():
    print_ltp(k, v, "A")
```
