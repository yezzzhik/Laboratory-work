```python
import sympy as sp

phi0, phi1, phi2, phi3, phi4, phi5 = sp.symbols('phi0 phi1 phi2 phi3 phi4 phi5')

i11, i22, i33, i44, i55 = sp.symbols('i11 i22 i33 i44 i55')

r1 = 110
r2 = 400
r3 = 40
r4 = 800
r5 = 300
r6 = 500
r7 = 300
r8 = 250

e1 = 20
e2 = 40

j1 = 10
j2 = 5

g1 = 1/r1
g2 = 1/r2
g3 = 1/r3
g4 = 1/r4
g5 = 1/r5
g6 = 1/r6
g7 = 1/r7
g8 = 1/r8


# Метод узловых потенциалов

eq1 = phi0

eq2 = phi1*(g1 + g7 + g8) - phi5*g1 - phi2*g7 - e1*g1

eq3 = phi2*(g6 + g7) - phi1*g7 - phi3*g6 + j2

eq4 = phi3*(g4 + g5 + g6) - phi4*g4 - phi2*g6

eq5 = phi4*(g3 + g4) - phi5*g3 - phi3*g4 - j1 + e2*g3

eq6 = phi5*(g1 + g2 + g3) - phi1*g1 - phi4*g3 - e2*g3 + e1*g1

sol_nodes = sp.solve(
    [eq1, eq2, eq3, eq4, eq5, eq6],
    [phi0, phi1, phi2, phi3, phi4, phi5],
    dict=True
)

print("Метод узловых потенциалов \n")

if sol_nodes:
    s = sol_nodes[0]

    phi0v = s[phi0]
    phi1v = s[phi1]
    phi2v = s[phi2]
    phi3v = s[phi3]
    phi4v = s[phi4]
    phi5v = s[phi5]

    i1 = (phi5v - phi1v + e1) / r1
    i2 = (phi5v - phi0v) / r2
    i3 = (-phi5v + phi4v + e2) / r3
    i4 = (phi4v - phi3v) / r4
    i5 = phi3v / r5
    i6 = (phi3v - phi2v) / r6
    i7 = (phi1v - phi2v) / r7
    i8 = phi1v / r8

    for var in [phi0, phi1, phi2, phi3, phi4, phi5]:
        print(f"{var} = {sp.N(s[var]):.6f} V")
    print()

    vals = [i1, i2, i3, i4, i5, i6, i7, i8]
    names = ['i1', 'i2', 'i3', 'i4', 'i5', 'i6', 'i7', 'i8']

    for n, v in zip(names, vals):
        print(f"{n} = {sp.N(v):.6f} A")


# Метод контурных токов

eqm1 = i11*(r2+r3) - i22*r2 + i55*(r4+r5) - i44*r5 - e2

eqm2 = i22*(r1+r2+r8) - i11*r2 - i33*r8 - e1

eqm3 = i33*(r7+r8) - i22*r8 + i44*(r5+r6) - i55*r5

eqm4 = -i55 + i11 - j1

eqm5 = i33 - i44 - j2

sol_mesh = sp.solve(
    [eqm1, eqm2, eqm3, eqm4, eqm5],
    [i11, i22, i33, i44, i55],
    dict=True
)

print("\n Метод контурных токов \n")

if sol_mesh:
    s = sol_mesh[0]

    i11v = s[i11]
    i22v = s[i22]
    i33v = s[i33]
    i44v = s[i44]
    i55v = s[i55]

    i1 = i22v
    i2 = i11v - i22v
    i3 = i11v
    i4 = i55v
    i5 = i44v - i55v
    i6 = -i44v
    i7 = i33v
    i8 = i22v - i33v

    for var in [i11, i22, i33, i44, i55]:
        print(f"{var} = {sp.N(s[var]):.6f} A")

    print()

    vals = [i1, i2, i3, i4, i5, i6, i7, i8]
    names = ['i1', 'i2', 'i3', 'i4', 'i5', 'i6', 'i7', 'i8']

    for n, v in zip(names, vals):
        print(f"{n} = {sp.N(v):.6f} A")
```
